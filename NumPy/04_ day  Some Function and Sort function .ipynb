Logical_Function

```python
import numpy as np
a=np.array([1,2,3,4,5,6])
b=np.array([7,8,9,10,11,12])           #index wise compare korbe
greater_than = a>b
print(greater_than)
```

```python
less_than = a<b
print(less_than)
```

```python
equal= a==b
print(equal)
```

Use_of_Any_&_All

```python
print(np.all(greater_than))
print(np.all(less_than))
print(np.any(equal))
```

Sorting

```python
p=np.array([11,43,5,64,45])
print("p was :",p)
z=p.copy()
print("copied z =" ,z)
z.sort()              # Inplase sort
print("sorted z =",z) # we could not recover past z values
```

```python
print(p)
sorted_arr=np.sort(p)
print(sorted_arr)
print(p)
```

2D _array_ SOrt

```python
mat=np.array([[6,4,5,7,9],[22,21,24,26,23]])
print("matrix was = \n", mat)



#horizontal sort :
hor_sort=np.sort(mat,axis=1)
print("\n\n horizontal , row wise sort : \n" ,hor_sort)


#vartical sort
var_sort=np.sort(mat,axis=0)
print("\n\n vartical /column wise sort : \n" ,var_sort)


```
