Logical_Function

```python
import numpy as np
a=np.array([1,2,3,4,5,6])
b=np.array([7,8,9,10,11,12])           #index wise compare korbe
greater_than = a>b
print(greater_than)
```

```python
less_than = a<b
print(less_than)
```

```python
equal= a==b
print(equal)
```

Use_of_Any_&_All

```python
print(np.all(greater_than))
print(np.all(less_than))
print(np.any(equal))
```

2D _array_ SOrt

```python
mat=np.array([[6,4,5,7,9],[22,21,24,26,23]])
print("matrix was = \n", mat)



#horizontal sort :
hor_sort=np.sort(mat,axis=1)
print("\n\n horizontal , row wise sort : \n" ,hor_sort)


#vartical sort
var_sort=np.sort(mat,axis=0)
print("\n\n vartical /column wise sort : \n" ,var_sort)


```

```python
#data loading ->
data=np.genformtxt('student_scores.csv',delimiter=',')
# by np.genformrxt we caan acess the txt and csv file from same directory
#we use delimiter becasuse the initial data/marks was divided with , like[72,54,77]
print(data)
```

```python
#statistics
math_marks=data[::,:1]
print(math_marks)

#maximum
max_math_marks= np.max(math_marks)
print(max_math_marks)

#minimum
min_math_marks= np.min(math_marks)
print(min_math_marks)

#avarage (mean)
avarage_math_marks= np.mean(math_marks)
print(avarage_math_marks)


#median
median_math_marks= np.median(math_marks)
print(median_math_marks)


#standard- diviation ( idicate how our data is divarse )
std_math_marks= np.std(math_marks)
print(std_math_marks)

```

cor_relation : np.corrcoef(data)

```python
study_times=np.array([2,3,4,5,6,7])
exam_marks=np.array([56,67,69,73,78,81])

data=np.array([study_times,exam_marks])
correlation=np.corrcoef(data)
print("Correlation marks :\n",correlation)

# How we can define the Relationship is good or bad : ->
# +0.90 to +1.00	Very Strong Positive
# +0.70 to +0.89	Strong Positive
# +0.40 to +0.69	Moderate Positive
# +0.20 to +0.39	Weak Positive
# 0	No Linear Correlation
# -0.20 to -0.39	Weak Negative
# -0.40 to -0.69	Moderate Negative
# -0.70 to -0.89	Strong Negative
# -0.90 to -1.00	Very Strong Negative
```
