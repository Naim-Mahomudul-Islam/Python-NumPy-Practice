8.6 -> np.arange

```python
import numpy as np
arr = np.arange(1,10,2)                        #[range from Start , end (excluded),difference]
print("So The ndArray is : ", arr)
print("Type :" , type(arr))
print("Data Type :" , arr.dtype)
print("shape  :" , arr.shape)
print('Dimention :' , arr.ndim,'D')
print("size:",arr.size)
```

arr = np.linespace(start, end (included), point will pick)

```python
import numpy as np
arr = np.linspace(1,10,15)                        #[range from Start , end (included),point]
print("So The ndArray is : ", arr)         #difference will same in every point
print("Type :" , type(arr))
print("Data Type :" , arr.dtype)
print("shape  :" , arr.shape)
print('Dimention :' , arr.ndim,'D')
print("size:",arr.size)
```

arr = np.logspace(start , end(included), point will pick ) #power wise 10^power

```python
import numpy as np
arr = np.logspace(0,4,5)                        #[range from Start , end (included),point]
print("So The ndArray is : ", arr)         #difference will same in every point
print("Type :" , type(arr))                     #10^0 10^1 10^2 10^3 10^4
print("Data Type :" , arr.dtype)
print("shape  :" , arr.shape)
print('Dimention :' , arr.ndim,'D')
print("size:",arr.size)
```

```python
import numpy as np
arr = np.logspace(0,4,5,base=2)                        #[range from Start , end (included),point]
print("So The ndArray is : ", arr)         #difference will same in every point
print("Type :" , type(arr))                     #2^0 2^1 2^2 2^3 2^4
print("Data Type :" , arr.dtype)
print("shape  :" , arr.shape)
print('Dimention :' , arr.ndim,'D')
print("size:",arr.size)
```

8.8 Advance Indexiing & Iteration

```python
import numpy as np
lst=[10,20,30,40,50]
arr5 = np.array(lst)
print(arr5)

arr5_new=arr5[1:4]
print(arr5_new)
arr5_new[1]=300
print(arr5_new)
print(arr5)
```

copy_index

```python
import numpy as np
lst=[10,20,30,40,50]
arr5 = np.array(lst)
print(arr5)

arr5_new=arr5[1:4].copy()                    #copy
print(arr5_new)
arr5_new[1]=300
print(arr5_new)
print(arr5)
```
