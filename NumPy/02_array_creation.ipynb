```python
import numpy as np
arr=np.array([1,2,3,4])
print(arr.dtype)
```

```python
import numpy as np
arr=np.array([1,2,3,4.3])
print(arr.dtype)
```

```python
import numpy as np
arr=np.array([1,2,3.8,'h','3'])  #int to sting maximum length 21
print(arr.dtype)                 #float to sting maximum length 32
```

```python
import numpy as np
arr=np.array([1,2,3,4],dtype=np.int8) #convert into int8 (int8 = range -127 to 127)
print(arr.dtype) #if we use more the 127 in give error
```

```python
arr=np.array([1,2,3.8])
print(arr.dtype)
arr=arr.astype(np.int16)              #astype
print(arr.dtype)
```
