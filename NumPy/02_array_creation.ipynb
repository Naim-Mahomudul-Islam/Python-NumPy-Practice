```python
dic = { 'a':10 ,  'b':20 , 'c':30 }
keys = dic.keys()
values=dic.values()
items=dic.items()
print(keys)
print(values)
print(items)
```

dictionary key in nd_array  :

```python
dic = { 'a':10 ,  'b':20 , 'c':30 }
keys = dic.keys()
values=dic.values()
items=dic.items()
arr= np.array(list(keys))
print("So The ndArray is : ", arr)
print("Type :" , type(arr))
print("Data Type :" , arr.dtype)
print("shape  :" , arr.shape)
print('Dimention :' , arr.ndim,'D')
print("size:",arr.size)
```

Dictionary values in nd_array :  

```python
dic = { 'a':10 ,  'b':20 , 'c':30 }
keys = dic.keys()
values=dic.values()
items=dic.items()
arr= np.array(list(values))
print("So The ndArray is : ", arr)
print("Data Type :" , arr.dtype)
arr = arr.astype(np.int16)
print("Type :" , type(arr))
print("Changed Data Type :" , arr.dtype)
print("shape :" , arr.shape)
print('Dimention :' , arr.ndim,'D')
print("size:",arr.size)
```
