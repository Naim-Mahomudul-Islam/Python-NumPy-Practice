```python
a=np.random.randint(1,10,size=(2,3))
b=np.random.randint(20,30,size=(2,3))
print(a)
print(b)
```

Concatenate

```python
con_row=np.concatenate((a,b),axis=0) #row_wise ->column same
print(con_row)

con_col=np.concatenate((a,b),axis=1) #column_wise ->row same
print(con_col)

```

Array Arithmetic operation

```python
#faster for Vatorization
x=np.array([10,20,30,40])
y=np.array([2,3,4,5])
add=x+y    # or we can use ->    np.add(a,b)
print(add)

sub=(x-y)
print(sub)

mul=(x*y)
print(mul)

div=x/y
print(div)

```

functional operation = trigonometry

```python
sin_val =np.sin(x)         #np.arcsin=> sin inverse
print(sin_val)         #sin cos tan give value in radian


#radian to degree @*180/^
deg_convertion=np.rad2deg(x)
print(deg_convertion)
```

```python
#Log

base_10_log_value=np.log10(x)
print(base_10_log_value)




#sqrut
print(x)
sqrt_val=np.sqrt(x)
print(sqrt_val)
```

some difference

```python
print(x)
total_sum=np.sum(x)
print(total_sum)

#cumulative_sum
print(x)
cumulative_sum=np.cumulative_sum(x) #kind of prefix sum
print(cumulative_sum)


```
